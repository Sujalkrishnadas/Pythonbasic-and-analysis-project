## 2. Tokenization
### 2.1 Word Tokenization
Using `word_tokenize` to split text into individual words (tokens).

```python
from nltk.tokenize import word_tokenize

# Example 1
text1 = "Hello, world! Welcome to NLP."
print("Example 1 Input:", text1)
tokens1 = word_tokenize(text1)  # Store tokens
print("Tokens:", tokens1)
print("Number of tokens:", len(tokens1))  # Print count

# Example 2: Handling contractions and punctuation
text2 = "Don't hesitate—tokenize this text, please."
print("\nExample 2 Input:", text2)
tokens2 = word_tokenize(text2)  # Store tokens
print("Tokens:", tokens2)
print("Number of tokens:", len(tokens2))  # Print count

# Example 3: Mixed language
text3 = "Bonjour! How are you today?"
print("\nExample 3 Input:", text3)
tokens3 = word_tokenize(text3)  # Store tokens
print("Tokens:", tokens3)
print("Number of tokens:", len(tokens3))  # Print count
```

### 2.2 Sentence Tokenization
Using `sent_tokenize` to split text into sentences.

```python
from nltk.tokenize import sent_tokenize

# Example 1
text1 = "This is sentence one. This is sentence two."
print("Example 1 Input:", text1)
print("Sentences:", sent_tokenize(text1))

# Example 2: Abbreviations
text2 = "Dr. Smith went to Washington. He arrived at 10 p.m."
print("\nExample 2 Input:", text2)
print("Sentences:", sent_tokenize(text2))

# Example 3: Emphatic punctuation
text3 = "Wow!!! That's amazing. Really, it is."
print("\nExample 3 Input:", text3)
print("Sentences:", sent_tokenize(text3))
```

## 3. Stemming
Using `PorterStemmer` to reduce words to their root forms by chopping off prefixes/suffixes.

```python
from nltk.stem import PorterStemmer

stemmer = PorterStemmer()

# Example 1: Regular verbs
words1 = ["running", "jumps", "easily"]
print("Example 1:", {w: stemmer.stem(w) for w in words1})

# Example 2: Irregular forms
words2 = ["studies", "studying", "studied"]
print("Example 2:", {w: stemmer.stem(w) for w in words2})

# Example 3: Mixed words
words3 = ["connections", "connecting", "connected"]
print("Example 3:", {w: stemmer.stem(w) for w in words3})
```

## 4. Lemmatization
Using `WordNetLemmatizer` to reduce words to their dictionary (base) forms with POS context.

```python
from nltk.stem import WordNetLemmatizer

lemmatizer = WordNetLemmatizer()

# Example 1: Verbs (specify POS)
print("running ->", lemmatizer.lemmatize("running", pos='v'))

# Example 2: Adjectives
print("better ->", lemmatizer.lemmatize("better", pos='a'))

# Example 3: Nouns
print("geese ->", lemmatizer.lemmatize("geese", pos='n'))
```

## 5. Stem vs. Lemma Comparison
Compare outputs of stemming vs. lemmatization.

```python
words = ["running", "caring", "happily", "geese", "better"]
for w in words:
    stem = stemmer.stem(w)
    lemma = lemmatizer.lemmatize(w)
    print(f"{w:10} -> stem: {stem:10} | lemma: {lemma}")
```

## 7. POS Tagging with Lemmatization
Use POS tags to improve lemmatization accuracy.

```python
from nltk.corpus import wordnet

def get_wordnet_pos(treebank_tag):
    if treebank_tag.startswith('J'): return wordnet.ADJ
    if treebank_tag.startswith('V'): return wordnet.VERB
    if treebank_tag.startswith('N'): return wordnet.NOUN
    if treebank_tag.startswith('R'): return wordnet.ADV
    return wordnet.NOUN

text = word_tokenize("The striped bats are hanging on their feet for best.")
tagged = pos_tag(text)

for word, tag in tagged:
    wn_tag = get_wordnet_pos(tag)
    lemma = lemmatizer.lemmatize(word, pos=wn_tag)
    print(f"{word:<10} POS:{tag:<5} -> Lemma:{lemma}")
```

## 8. Stop-word Removal
Filter out common stop words to focus on content words.

```python
from nltk.corpus import stopwords
stop_words = set(stopwords.words('english'))
text1 = "This is an example sentence with some stop words. hey this is me"
words = word_tokenize(text1)
print(words)
filtered_words = [word for word in words if word.lower() not in stop_words]
print(filtered_words)
```

## 2. Tokenization
spaCy provides tokenization out-of-the-box, handling punctuation, contractions, and special cases.

```python
# Example 1: Basic word tokens
doc = nlp("Hello, world! Welcome to spaCy.")
print([token.text for token in doc])
print("Number of tokens:", len(doc))  # Added line

# Example 2: Contractions
doc = nlp("Don't hesitate to use spaCy's tokenizer.")
print([token.text for token in doc])
print("Number of tokens:", len(doc))  # Added line

# Example 3: URLs and emails
doc = nlp("Contact us at support@example.com or visit https://example.com/page.")
print([token.text for token in doc])
print("Number of tokens:", len(doc))  # Added line
```

## 3. Named Entity Recognition (NER)
spaCy's pretrained model recognizes entities like PERSON, ORG, GPE, etc.

```python
# Example 1: Simple sentence
doc = nlp("Apple is looking at buying U.K. startup for $1 billion.")
for ent in doc.ents:
    print(ent.text, ent.label_)

# Explanation:
# Recognizes entities such as ORG, GPE, MONEY.

# Example 2: Date and events
doc = nlp("The conference will be held on July 10, 2025 in New York.")
print([(ent.text, ent.label_) for ent in doc.ents])

# Explanation:
# Dates and locations are identified.

# Example 3: Multiple entities
text = "Barack Obama and Michelle Obama visited Microsoft and Google in Seattle."
doc = nlp(text)
print([(ent.text, ent.label_) for ent in doc.ents])

# Explanation:
# Recognizes multiple PERSON and ORG entities.
```

## 4. Part-of-Speech (POS) Tagging
spaCy provides fine-grained POS tags and universal tags.

```python
# Example 1: Simple sentence
doc = nlp("She sells seashells by the seashore.")
print([(token.text, token.pos_, token.tag_) for token in doc])

# Explanation:
# pos_ is the coarse POS, tag_ is detailed tag.

# Example 2: Questions
doc = nlp("What time is the meeting tomorrow?")
print([(token.text, token.pos_, token.tag_) for token in doc])

# Explanation:
# Helps with understanding grammatical structure.

# Example 3: Complex sentence
doc = nlp("The quick brown fox jumps over the lazy dog.")
print([(token.text, token.pos_, token.tag_) for token in doc])

# Explanation:
# Demonstrates tags for adjectives, nouns, verbs, etc.
```

## 5. Dependency Parsing
Dependency parsing reveals the grammatical relationships between tokens.

```python
# Example 1: Basic parse
doc = nlp("The cat sat on the mat.")
for token in doc:
    print(token.text, token.dep_, token.head.text)

# Explanation:
# dep_ shows relation; head.text shows governor.

# Example 2: Relative clauses
doc = nlp("The man who wore a hat sat down.")
print([(token.text, token.dep_, token.head.text) for token in doc])

# Explanation:
# 'who' links clause to 'man'.

# Example 3: Complex sentence
doc = nlp("Despite the rain, the match continued without interruption.")
print([(token.text, token.dep_, token.head.text) for token in doc])

# Explanation:
# Handles prepositional attachments and conjunctions.
```

## 6. Custom Pipeline Component
Add a simple custom component that prints token text and POS.

```python
import spacy
from spacy.language import Language

# Define custom component with the @Language.component decorator
@Language.component("show_token_info")  # Register the component with a name
def show_token_info(doc):
    print("Custom component output:")
    for token in doc[:5]:
        print(token.text, token.pos_)
    return doc

# Load the English model
nlp = spacy.load("en_core_web_sm")

# Add to pipeline using the registered name
nlp.add_pipe("show_token_info", first=True)

# Run on example text
doc = nlp("SpaCy pipelines are easy to customize.")

# Remove component to avoid duplicate prints later
nlp.remove_pipe("show_token_info")
```

## 7. Similarity
Compute similarity scores between documents or tokens using word vectors.

```python
# Example 1: Token similarity
import spacy
nlp = spacy.load("en_core_web_sm")
doc = nlp("cat dog banana")
print(doc[0].similarity(doc[1]))  # cat vs dog

# Explanation:
# Returns a float similarity score.

# Example 2: Document similarity
doc1 = nlp("Bengaluru")
doc2 = nlp("Bangalore")
print(doc1.similarity(doc2))

# Explanation:
# Measures semantic similarity between docs.

# Example 3: Word vs phrase similarity
token = nlp("Paris")[0]
phrase = nlp("the city of love")
print(token.similarity(phrase[2]))

# Explanation:
# Even single tokens vs phrases can be compared.
```
